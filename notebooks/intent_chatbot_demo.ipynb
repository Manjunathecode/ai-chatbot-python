# Intent-Based Support Chatbot Demo

This output-free notebook loads the same TF-IDF + MLP classifier used by the Flask application. Use fictional prompts only; the project is not connected to any real support system.

```python
from pathlib import Path
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))

from chatbot import IntentChatbot

bot = IntentChatbot(project_root / 'intents.json')
bot.intent_count, bot.pattern_count
```

```python
demo_prompts = [
    'Where is my order?',
    'My payment failed',
    'How do I reset my password?',
    'The app is not working',
]

for prompt in demo_prompts:
    prediction = bot.predict(prompt)
    print(prompt, '->', prediction.intent, f'{prediction.confidence:.1%}')
```

## Interpretation boundary

The confidence is the classifier's probability estimate over the small curated intent catalog. It is not a measure of factual correctness, service availability, or production reliability.
